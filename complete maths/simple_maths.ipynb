Create Database

```python
import sqlite3
import pandas as pd

# Create or connect to a database file
conn = sqlite3.connect("shop.db")
cur = conn.cursor()

print("Database connected successfully.")
```

Create simple Table

```python
cur.execute("""
CREATE TABLE IF NOT EXISTS sales (
    sale_id INTEGER PRIMARY KEY AUTOINCREMENT,
    product_name TEXT,
    category TEXT,
    price REAL,
    quantity INTEGER
);
""")

conn.commit()
print("Table created.")
```

Insert sample data

```python
cur.executemany("""
INSERT INTO sales (product_name, category, price, quantity)
VALUES (?, ?, ?, ?);
""", [
    ('Laptop', 'Electronics', 800, 2),
    ('Phone', 'Electronics', 500, 3),
    ('Headphones', 'Electronics', 100, 5),
    ('Desk Chair', 'Furniture', 150, 4),
    ('Desk', 'Furniture', 300, 2),
    ('Notebook', 'Stationery', 5, 20),
    ('Pen Pack', 'Stationery', 10, 10)
])

conn.commit()
print("Data inserted.")
```

View all data

```python
import pandas as pd

# Query all data from the 'sales' table into a DataFrame
df = pd.read_sql_query("SELECT * FROM sales", conn)

# Display the DataFrame
df
```

Calculate the AVG Price

```python
# Calculate the average price using SQL
df_avg = pd.read_sql_query("SELECT AVG(price) AS average_price FROM sales", conn)

print(df_avg)
```

Calculate Total Quantity Sold

```python
# Calculate total quantity sold using SQL
df_total_qty = pd.read_sql_query("SELECT SUM(quantity) AS total_quantity_sold FROM sales", conn)

print(df_total_qty)
```

Calculate Highest and Lowest Price

```python
# Highest price
df_max_price = pd.read_sql_query("SELECT MAX(price) AS max_price FROM sales", conn)

# Lowest price
df_min_price = pd.read_sql_query("SELECT MIN(price) AS min_price FROM sales", conn)

# Both in a single query
df_price_range = pd.read_sql_query("""
SELECT
    MAX(price) AS max_price,
    MIN(price) AS min_price
FROM sales
""", conn)

print(df_price_range)
```

*Calculate* revenue per product





```python
df_revenue_per_product = pd.read_sql_query("""
SELECT
    product_name,
    price,
    quantity,
    (price * quantity) AS total_revenue
FROM sales
ORDER BY total_revenue DESC
""", conn)

print(df_revenue_per_product)
```



Calculate total revenue

```python
df_total_revenue = pd.read_sql_query("""
SELECT SUM(price * quantity) AS grand_total_revenue
FROM sales
""", conn)

print(df_total_revenue)
```

Calculate revenue per category

```python
df_revenue_per_category = pd.read_sql_query("""
SELECT
    category,
    SUM(price * quantity) AS category_revenue
FROM sales
GROUP BY category
ORDER BY category_revenue DESC
""", conn)

print(df_revenue_per_category)
```

Calculate average price per category

```python
df_avg_price_per_category = pd.read_sql_query("""
SELECT
    category,
    AVG(price) AS average_price
FROM sales
GROUP BY category
""", conn)

print(df_avg_price_per_category)
```

Calculate count products per category

```python
df_product_count = pd.read_sql_query("""
SELECT
    category,
    COUNT(product_name) AS product_count
FROM sales
GROUP BY category
""", conn)

print(df_product_count)
```

Which category has a revenue greater than 2000

```python
df_high_revenue_categories = pd.read_sql_query("""
SELECT
    category,
    SUM(price * quantity) AS total_revenue
FROM sales
GROUP BY category
HAVING SUM(price * quantity) > 2000
""", conn)

print(df_high_revenue_categories)
```
